```python
from maxent_insect_dir import config
import os
import pathlib
import geopandas as gpd
import rasterio
import rasterio.plot as rioplot # To Visualize Raster Data
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd


pd.options.mode.chained_assignment = None  # default='warn'
```

```python
# Define the directory containing your WorldClim TIFF files
tiff_dir = config.config_['climate_data']

# List of WorldClim files you want to visualize
tiff_files = os.listdir(config.config_['climate_data'])

ignore_files = ['readme.md']

# Function to visualize a single TIFF file
def visualize_tiff(file_path, title):
    with rasterio.open(file_path) as dataset:
        data = dataset.read(1)  # Read the first band
        print(f'title : {title}, data.shape : {data.shape}')
        # plt.figure(figsize=(8, 6))
        # plt.imshow(data, cmap='viridis')
        # plt.colorbar(label='Value')
        # plt.title(title)
        # plt.axis('off')
        # plt.show()

#Loop through each file and visualize it
for tiff_file in tiff_files:
    if tiff_file not in ignore_files:
        file_path = os.path.join(tiff_dir, tiff_file)
        title = os.path.splitext(tiff_file)[0]  # Use filename (without extension) as title
        visualize_tiff(file_path, title)
```
